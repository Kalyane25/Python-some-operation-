# Ensemble Learning Assignment
This notebook contains answers to both theoretical and practical questions related to Bagging, Random Forest, and other ensemble techniques.

## 1. Can we use Bagging for regression problems?

Yes, Bagging can be used for regression problems. The base learners are trained on different subsets of the data and their predictions are averaged for the final output.

## 2. What is the difference between multiple model training and single model training?

Single model training involves training one model on the entire dataset. Multiple model training (ensemble learning) involves training multiple models and combining their predictions to improve performance.

## 3. Explain the concept of feature randomness in Random Forest

Random Forest introduces feature randomness by selecting a random subset of features at each split, promoting model diversity and reducing overfitting.

## 4. What is OOB (Out-of-Bag) Score

OOB score is an internal validation method in bagging. It uses the samples not selected in the bootstrap to test the model trained on the selected samples.

## 5. How can you measure the importance of features in a Random Forest model?

Feature importance is measured by observing the average decrease in impurity or the increase in model error when a feature is permuted.

## 6. Explain the working principle of a Bagging Classifier

A Bagging Classifier trains multiple instances of a base estimator on different bootstrap samples and combines their predictions via majority voting.

## 7. How do you evaluate a Bagging Classifier’s performance

Use classification metrics like accuracy, precision, recall, F1-score, and ROC-AUC on a test or validation set.

## 8. How does a Bagging Regressor work

A Bagging Regressor trains multiple regressors on different subsets and averages their predictions.

## 9. What is the main advantage of ensemble techniques

They reduce variance (bagging), bias (boosting), and improve predictive performance.

## 10. What is the main challenge of ensemble methods

They are more complex, require more resources, and are harder to interpret.

## 11. Explain the key idea behind ensemble techniques

Combine multiple models to produce a more robust and accurate prediction.

## 12. What is a Random Forest Classifier

It is an ensemble of decision trees using bagging and random feature selection.

## 13. What are the main types of ensemble techniques

Bagging, Boosting, and Stacking.

## 14. What is ensemble learning in machine learning

It refers to techniques that create multiple models and combine them to solve a problem.

## 15. When should we avoid using ensemble methods

When the dataset is small or the model interpretability is crucial.

## 16. How does Bagging help in reducing overfitting

By averaging predictions from multiple models trained on different samples, it reduces variance.

## 17. Why is Random Forest better than a single Decision Tree

Because it reduces overfitting and provides better generalization.

## 18. What is the role of bootstrap sampling in Bagging

It ensures diversity among base learners by training each on a different subset.

## 19. What are some real-world applications of ensemble techniques

Fraud detection, recommendation systems, medical diagnosis, and financial modeling.

## 20. What is the difference between Bagging and Boosting?

Bagging builds models independently and combines them (reduces variance), Boosting builds sequentially with focus on errors (reduces bias).
